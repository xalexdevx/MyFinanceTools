Valuation: A Conceptual Overview:

In this chapter, therefore, you will find a detailed overview of the core building blocks of the valuation techniques used by investment bankers.



Enterprise Value (EV) represents the total value of a company’s operations. It is different from Market Value (or Equity Value), which only reflects the value belonging to shareholders after other stakeholders’ claims are deducted.

Because of this, EV adjusts Market Value to include all parts of the business, not just equity. In other words, EV bridges the gap between what shareholders own and what the entire business is worth.

$$
\text{Enterprise Value}
= \text{Operating Value of the Company}
= \text{Net Value of all Claims on the Company’s Assets (Excluding Excess Cash)}
$$

This equation is simple enough. Assuming that the company is profitable, and that it’s more valuable in operation than in liquidation (in other words, the company is worth more money if it continues in business rather than stopping business and selling off the assets to the highest bidders), the value of the company is equal to the value of its productive operations. This value must also equal the value of all of the net claims against the company’s assets, because these assets are being used to produce money for these claimants, or stakeholders (assuming, of course, that these claims have been valued correctly). The “Excluding Excess Cash” piece will be explained in a moment.

The net value of all the claims on a company’s assets can be broken down as follows:

* **Debt**:
  Money that has been lent to the company by another person or institution. Debt holders have a higher priority than equity holders on the claims of the company’s assets and value, so they get paid first. To arrive at Enterprise Value (EV), Debt is added to the Market Value of Equity.

* **Cash**:
  Money owned by the company and shown on its balance sheet. If this cash is not required for operations, it could be used to pay down Debt or repurchase Equity. Therefore, the more excess Cash a company holds, the less valuable its operations must be (for a fixed Equity Value). To arrive at EV, Cash is subtracted from the Market Value of Equity. In practice, Cash is often netted against Debt to compute **Net Debt**.

* **Minority Interest**:
  Represents the portion of a consolidated subsidiary that is owned by outside shareholders. Although the subsidiary is fully consolidated, the parent does not own 100% of it, so this claim must be treated like Debt. Minority Interest is therefore added to the Market Value of Equity when calculating EV.

* **Preferred Equity**:
  A hybrid security that is senior to Common Equity but junior to Debt, and typically pays fixed dividends. Because it behaves economically like Debt unless converted into Common Equity, Preferred Equity is added to the Market Value of Common Equity in the calculation of EV.
